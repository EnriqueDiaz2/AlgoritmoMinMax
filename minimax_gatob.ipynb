```python
# El tablero es una lista de 9 posiciones (0 a 8).
# " " = vacío, "X" = humano, "O" = computadora
#
#  0 | 1 | 2
#  3 | 4 | 5
#  6 | 7 | 8

tablero = [" "] * 9

def mostrar_tablero(tablero):
    print()
    print(f" {tablero[0]} | {tablero[1]} | {tablero[2]} ")
    print("---+---+---")
    print(f" {tablero[3]} | {tablero[4]} | {tablero[5]} ")
    print("---+---+---")
    print(f" {tablero[6]} | {tablero[7]} | {tablero[8]} ")
    print()

mostrar_tablero(tablero)
```

```python
# Todas las combinaciones que dan victoria (filas, columnas, diagonales)
COMBINACIONES_GANADORAS = [
    [0, 1, 2], [3, 4, 5], [6, 7, 8],   # filas
    [0, 3, 6], [1, 4, 7], [2, 5, 8],   # columnas
    [0, 4, 8], [2, 4, 6]               # diagonales
]

def hay_ganador(tablero, jugador):
    # Revisa si el jugador tiene 3 símbolos en alguna combinación
    for a, b, c in COMBINACIONES_GANADORAS:
        if tablero[a] == jugador and tablero[b] == jugador and tablero[c] == jugador:
            return True
    return False

def tablero_lleno(tablero):
    # Si ya no hay espacios vacíos, el juego terminó
    return " " not in tablero

def movimiento_valido(tablero, posicion):
    # Válido si está entre 0 y 8 y la casilla está vacía
    return 0 <= posicion <= 8 and tablero[posicion] == " "

def juego_terminado(tablero):
    return hay_ganador(tablero, "X") or hay_ganador(tablero, "O") or tablero_lleno(tablero)
```

```python
def minimax(tablero, es_turno_computadora):
    # Casos base: el juego terminó, devolvemos un puntaje
    if hay_ganador(tablero, "O"):
        return 1      # gana la computadora
    if hay_ganador(tablero, "X"):
        return -1     # gana el humano
    if tablero_lleno(tablero):
        return 0      # empate

    if es_turno_computadora:
        # La computadora (O) quiere el puntaje MÁS ALTO (maximiza)
        mejor_puntaje = -2
        for i in range(9):
            if tablero[i] == " ":
                tablero[i] = "O"                       # probar movimiento
                puntaje = minimax(tablero, False)      # ver qué pasaría
                tablero[i] = " "                       # deshacer movimiento
                mejor_puntaje = max(mejor_puntaje, puntaje)
        return mejor_puntaje
    else:
        # El humano (X) quiere el puntaje MÁS BAJO (minimiza)
        mejor_puntaje = 2
        for i in range(9):
            if tablero[i] == " ":
                tablero[i] = "X"
                puntaje = minimax(tablero, True)
                tablero[i] = " "
                mejor_puntaje = min(mejor_puntaje, puntaje)
        return mejor_puntaje

def mejor_movimiento(tablero):
    # Prueba cada casilla vacía y elige la de mayor puntaje
    mejor_puntaje = -2
    mejor_pos = None
    for i in range(9):
        if tablero[i] == " ":
            tablero[i] = "O"
            puntaje = minimax(tablero, False)   # ahora le toca al humano
            tablero[i] = " "
            if puntaje > mejor_puntaje:
                mejor_puntaje = puntaje
                mejor_pos = i
    return mejor_pos
```

```python

def pedir_movimiento(tablero):
    # Pide una casilla al humano hasta que sea válida
    while True:
        entrada = input("Tu turno (X). Elige una casilla del 1 al 9: ")
        if not entrada.isdigit():
            print("Escribe un número del 1 al 9.")
            continue
        posicion = int(entrada) - 1       # el usuario ve 1-9, la lista usa 0-8
        if movimiento_valido(tablero, posicion):
            return posicion
        print("Movimiento inválido. Casilla ocupada o fuera del tablero.")

def jugar():
    tablero = [" "] * 9
    print("¡Juego del Gato! Tú eres X, la computadora es O.")
    print("Casillas:\n 1 | 2 | 3\n 4 | 5 | 6\n 7 | 8 | 9")

    turno_humano = True    # el humano empieza
    mostrar_tablero(tablero)

    while not juego_terminado(tablero):
        if turno_humano:
            pos = pedir_movimiento(tablero)
            tablero[pos] = "X"
        else:
            print("Turno de la computadora (O)...")
            pos = mejor_movimiento(tablero)
            tablero[pos] = "O"

        mostrar_tablero(tablero)
        turno_humano = not turno_humano     # alternar turnos

    if hay_ganador(tablero, "X"):
        print("¡Ganaste!")
    elif hay_ganador(tablero, "O"):
        print("La computadora gana.")
    else:
        print("¡Empate!")

jugar()
```

```python
def probar_todas(tablero, turno_humano):
    # Devuelve cuántas veces pierde la computadora (debe ser 0)
    if hay_ganador(tablero, "X"):
        return 1
    if hay_ganador(tablero, "O") or tablero_lleno(tablero):
        return 0
    derrotas = 0
    if turno_humano:
        for i in range(9):
            if tablero[i] == " ":
                tablero[i] = "X"
                derrotas += probar_todas(tablero, False)
                tablero[i] = " "
    else:
        pos = mejor_movimiento(tablero)
        tablero[pos] = "O"
        derrotas += probar_todas(tablero, True)
        tablero[pos] = " "
    return derrotas

print("Derrotas de la computadora:", probar_todas([" "] * 9, True))
```
